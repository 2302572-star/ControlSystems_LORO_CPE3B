```python
import os
from docx import Document
from openai import OpenAI

# =====================================
# LM STUDIO CONNECTION
# =====================================
client = OpenAI(
    base_url="http://127.0.0.1:1234/v1",
    api_key="lm-studio"
)

# =====================================
# SAVE LOCATION (DIRECTLY TO DESKTOP)
# =====================================
DOCS_FOLDER = os.path.join(os.path.expanduser("~"), "Desktop")

print(f"Documents will be saved to:\n{DOCS_FOLDER}")

# =====================================
# STATUS DISPLAY
# =====================================
def show_status(action, details=""):
    if details:
        print(f"\n[{action}] User: {details}")
    else:
        print(f"\n[{action}]")

# =====================================
# SAVE DOCUMENT
# =====================================
def save_document(topic, content):
    safe_name = "".join(
        c for c in topic if c.isalnum() or c in (" ", "_", "-")
    ).strip()

    filename = safe_name.replace(" ", "_") + ".docx"
    filepath = os.path.join(DOCS_FOLDER, filename)

    doc = Document()
    doc.add_heading(topic.title(), level=1)
    doc.add_paragraph(content)
    doc.save(filepath)

    return filepath

# =====================================
# AI REQUEST
# =====================================
def ask_ai(system_prompt, user_prompt):
    response = client.chat.completions.create(
        model="qwen/qwen3-1.7b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.7,
        max_tokens=2000
    )

    return response.choices[0].message.content

# =====================================
# DOCUMENT AGENT
# =====================================
def document_agent(user_input):
    try:
        lower_input = user_input.lower()

        start = lower_input.find("save the") + len("save the")
        end = lower_input.find("as docs")

        topic = user_input[start:end].strip()

        if not topic:
            print("Bot: Please use:")
            print("save the <topic> as docs")
            return

        print(f"\nBot: Generating detailed document about '{topic}'...")

        content = ask_ai(
            "You are an expert researcher. Write detailed, well-structured documents with multiple paragraphs.",
            f"""
            Create a detailed report about:
            {topic}

            Include:
            - Introduction
            - Main Concepts
            - Applications
            - Advantages
            - Challenges
            - Conclusion
            """
        )

        if not content:
            content = "No content generated."

        print("Bot: Creating Word document...")

        filepath = save_document(topic, content)

        print("\nBot: Document saved successfully!")
        print(f"Bot: Location -> {filepath}")

        # OPEN DOCUMENT AUTOMATICALLY
        os.startfile(filepath)

        print("Bot: Document opened automatically.")

    except Exception as e:
        print("\nBot: Error generating document.")
        print("Details:", e)

# =====================================
# CHAT AGENT
# =====================================
def chat_agent(user_input):
    try:
        response = ask_ai(
            "You are a helpful AI assistant.",
            user_input
        )

        print(f"\nBot: {response}")

    except Exception as e:
        print("\nBot: Error connecting to LM Studio.")
        print("Details:", e)

# =====================================
# MAIN CHATBOT
# =====================================
def chatbot():
    print("=" * 45)
    print("      AI CHATBOT WITH DOC AGENT")
    print("=" * 45)

    print("\nCommands:")
    print("save the <topic> as docs")
    print("exit")

    while True:
        try:
            user_input = input("\nUser: ").strip()

            if not user_input:
                continue

            if user_input.lower() == "exit":
                show_status("EXIT")
                print("Bot: Goodbye!")
                break

            elif (
                "save the" in user_input.lower()
                and "as docs" in user_input.lower()
            ):
                show_status("SAVE DOCUMENT", user_input)
                document_agent(user_input)

            else:
                show_status("CHAT MODE", user_input)
                chat_agent(user_input)

        except KeyboardInterrupt:
            print("\nBot: Goodbye!")
            break

        except EOFError:
            print("\nBot: Goodbye!")
            break

# =====================================
# RUN
# =====================================
if __name__ == "__main__":
    chatbot()
```
